```python
query = """ 
SELECT Canceled_Rides_by_Customer,
COUNT(*) AS Cancelled_by_Customers
FROM ola_data
GROUP BY Canceled_Rides_by_Customer
ORDER BY Cancelled_by_Customers DESC;
"""
df3 = pd.read_sql(query, con=engine)
df3
```

```python
query = """ 
SELECT Canceled_Rides_by_Driver,
COUNT(*) AS Cancelled_by_Driver
FROM ola_data
GROUP BY Canceled_Rides_by_Driver
ORDER BY Cancelled_by_Driver DESC;
"""
df51 = pd.read_sql(query, con=engine)
df51.iloc[1:]
```

```python
colors = ['red' if str(reason).strip() == 'Personal & Car related issue' else 'skyblue'
          for reason in df51['Canceled_Rides_by_Driver']]
df51plot = df51.iloc[1:].plot(kind='bar', x='Canceled_Rides_by_Driver', y='Cancelled_by_Driver', figsize=(6,6), color=colors)
plt.xlabel('Canceled Rides by Driver')
plt.ylabel('Cancelled Rides by Driver')
plt.title('Distribution of Canceled Rides by Driver')
plt.show()
```
